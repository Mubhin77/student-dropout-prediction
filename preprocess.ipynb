
# Stage 3: Feature Engineering and Preprocessing

The dataset contains both categorical and numerical
features.

Categorical variables were transformed using
OneHotEncoder. This was necessary because many of the
categorical variables were represented by numerical
category codes.

Numerical variables were standardised using
StandardScaler to bring them to a comparable scale.

A ColumnTransformer was used to apply the appropriate
transformation to each feature group.

The preprocessing operations were implemented in
src/features.py.

To prevent data leakage, the preprocessing object
was fitted only on the training dataset. The fitted
transformations were then applied to the testing dataset.

The resulting preprocessing component can be reused
during model training and deployment.
